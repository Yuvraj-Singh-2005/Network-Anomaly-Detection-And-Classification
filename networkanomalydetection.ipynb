```python
def countduplicate(df):
    duplicate_count=df.duplicated().sum()
    print(f" The number of the duplicated rows in the dataset is  {duplicate_count} amounting to the {(duplicate_count/len(df))*100 : .2f}% of total")
```

```python
df.drop_duplicates(inplace=True)
```

```python
le=LabelEncoder()
x=df.drop('Label',axis=1)
y=df['Label']
```

```python
y=le.fit_transform(y)
```

```python
from collections import Counter
```

```python
c=Counter(y)
for k in c.keys():
  print(k, c[k])
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42, stratify=y)
```

```python
train_class_counts = Counter(y_train)
print("Original training class distribution:", train_class_counts)
sampling_strategy = {}
for class_label, count in train_class_counts.items():
    if count < 100:
        sampling_strategy[class_label] = count * 30
adasyn = ADASYN(sampling_strategy=sampling_strategy, random_state=42, n_neighbors=5)
x_resampled, y_resampled = adasyn.fit_resample(x_train, y_train)
print("Resampled class distribution:", Counter(y_resampled))
```

```python
del y_train
```

```python
y_train = y_resampled
```

```python
del x_train
```

```python
x_train=x_resampled
```

```python
y_train=le.inverse_transform(y_train)
```

```python
c2=Counter(y_train)
print(c2)
```

```python
c3=Counter(y_test)
print(c3)
```

```python
y_test=le.inverse_transform(y_test)
```

```python
c4=Counter(y_test)
print(c4)
```

```python
label_mapping = {label: idx for idx, label in enumerate(le.classes_)}
print(label_mapping)
```

```python
rf = RandomForestClassifier(random_state=42, n_jobs=-1)
rf.fit(x_train, y_train)
```

```python
y_pred = rf.predict(x_test)
y_proba = rf.predict_proba(x_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')
class_report = classification_report(y_test, y_pred)
auc_roc = roc_auc_score(y_test, y_proba, multi_class='ovr')
```

```python
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
print("AUC-ROC Score:", auc_roc)
print("\nClassification Report:\n", class_report)
```

```python
y_train_enc = le.transform(y_train)
y_test_enc = le.transform(y_test)
```

```python
xgb_model = XGBClassifier(
    objective='multi:softprob',
    num_class=len(le.classes_),
    use_label_encoder=False,
    eval_metric='mlogloss',
    n_jobs=-1,
    random_state=42
)

xgb_model.fit(x_train, y_train_enc)
```

```python
y_pred_enc = xgb_model.predict(x_test)
y_pred = le.inverse_transform(y_pred_enc)

y_proba = xgb_model.predict_proba(x_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision (macro):", precision_score(y_test, y_pred, average='macro'))
print("Recall (macro):", recall_score(y_test, y_pred, average='macro'))
print("F1 Score (macro):", f1_score(y_test, y_pred, average='macro'))
print("AUC-ROC Score:", roc_auc_score(y_test_enc, y_proba, multi_class='ovr'))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(x_train, y_train_enc)

y_pred_dt = dt_model.predict(x_test)
y_pred_dt_decoded = le.inverse_transform(y_pred_dt)

accuracy = accuracy_score(y_test, y_pred_dt_decoded)
classification_rep = classification_report(y_test, y_pred_dt_decoded)
conf_matrix = confusion_matrix(y_test, y_pred_dt_decoded)
precision = precision_score(y_test, y_pred_dt_decoded, average='macro', zero_division=0)
recall = recall_score(y_test, y_pred_dt_decoded, average='macro', zero_division=0)
y_prob_dt = dt_model.predict_proba(x_test)
auc_roc = roc_auc_score(y_test_enc, y_prob_dt, multi_class='ovr')

print(f"Accuracy: {accuracy}")
print("\nClassification Report (Macro):")
print(classification_rep)
print(f"\nPrecision (Macro): {precision}")
print(f"Recall (Macro): {recall}")
print(f"AUC-ROC Score: {auc_roc}")
```

```python
from sklearn.ensemble import VotingClassifier
voting_clf = VotingClassifier(
    estimators=[('xgb', xgb_model), ('rf', rf)],
    voting='soft',
    n_jobs=-1
)

voting_clf.fit(x_train, y_train_enc)

y_pred_enc = voting_clf.predict(x_test)
y_pred_proba = voting_clf.predict_proba(x_test)

y_pred = le.inverse_transform(y_pred_enc)
y_test_decoded = le.inverse_transform(y_test_enc)

accuracy = accuracy_score(y_test_decoded, y_pred)
precision = precision_score(y_test_decoded, y_pred, average='macro', zero_division=0)
recall = recall_score(y_test_decoded, y_pred, average='macro', zero_division=0)
f1 = f1_score(y_test_decoded, y_pred, average='macro', zero_division=0)
auc_roc = roc_auc_score(y_test_enc, y_pred_proba, multi_class='ovr')

print("Ensemble Model Performance (Soft Voting):")
print(f"Accuracy: {accuracy}")
print(f"Precision (macro): {precision}")
print(f"Recall (macro): {recall}")
print(f"F1 Score (macro): {f1}")
print(f"AUC-ROC Score: {auc_roc}\n")
print("Classification Report:\n")
print(classification_report(y_test_decoded, y_pred))
```
